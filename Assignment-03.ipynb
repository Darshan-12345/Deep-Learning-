# Build the Artificial Neural Network (ANN)

An Artificial Neural Network (ANN) is composed of multiple layers of neurons.

Architecture used in this project:

- Input Layer
- Flatten Layer
- Hidden Layer 1 (128 Neurons, ReLU)
- Hidden Layer 2 (64 Neurons, ReLU)
- Output Layer (10 Neurons, Softmax)

### Forward Propagation

During forward propagation:

1. Input image is passed into the network.
2. Hidden layers compute weighted sums.
3. ReLU activation introduces non-linearity.
4. Output layer predicts probabilities for each class using Softmax.

### Backpropagation

During training:

- Prediction error (loss) is calculated.
- Gradients are computed.
- Weights and biases are updated using the optimizer.

## Create the Neural Network

We use TensorFlow's Sequential API to stack layers one after another.

## Create Model Function

A separate function is used to create a new neural network for each experiment. This ensures that each model starts with randomly initialized weights, making the comparison fair.

```python
def create_model(optimizer):
    model = Sequential([
        Flatten(input_shape=(28,28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])
    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model
```

```python
# Build the ANN model

model = Sequential([

    # Convert 28x28 image into a 784-element vector
    Flatten(input_shape=(28,28)),

    # Hidden Layer 1
    Dense(128, activation='relu'),

    # Hidden Layer 2
    Dense(64, activation='relu'),

    # Output Layer
    Dense(10, activation='softmax')
])
```

```python
for i, layer in enumerate(model.layers):
    print(f"Layer {i+1}")
    print("-----------------------")
    print("Name       :", layer.name)
    print("Type       :", layer.__class__.__name__)

    if hasattr(layer, "activation"):
        print("Activation :", layer.activation.__name__)

    print()
```

# Forward Propagation Analysis

Forward Propagation is the process of passing the input through the neural network to generate predictions.

In this step:

1. The input image is flattened into a 784-dimensional vector.
2. The data passes through hidden layers.
3. ReLU activation extracts meaningful patterns.
4. The output layer applies Softmax to produce class probabilities.
5. The class with the highest probability is selected as the prediction.

Since the model has not been trained yet, the predictions will be almost random. This demonstrates why training and backpropagation are necessary.

```python
# Display one sample image

sample_image = X_test[0]

plt.figure(figsize=(4,4))
plt.imshow(sample_image, cmap='gray')
plt.title(f"Actual Label: {class_names[y_test[0]]}")
plt.axis('off')
plt.show()
```

## Forward Pass

The sample image is passed through the neural network. The output layer returns the probability of the image belonging to each clothing category.

## Observation

# Backpropagation Analysis

Backpropagation is the process through which the neural network learns from its mistakes.

The steps are:

1. Forward propagation generates predictions.
2. The loss function measures the prediction error.
3. Gradients are computed for every trainable parameter.
4. The optimizer updates the weights and biases.
5. The process repeats for each epoch until the loss decreases.

```python
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
initial_weights = model.layers[1].get_weights()[0].copy()

print("Initial Weights Shape:", initial_weights.shape)

print(initial_weights[:5, :5])
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
updated_weights = model.layers[1].get_weights()[0]

print(updated_weights[:5, :5])
```

```python
weight_difference = updated_weights - initial_weights

print(weight_difference[:5, :5])
```

## Compare Weight Changes

The following code calculates the average change in the weights after one training epoch.

```python
average_change = np.mean(np.abs(weight_difference))

print(f"Average Weight Change: {average_change:.6f}")
```

```python
print("Loss after one epoch:", history.history["loss"][0])
print("Accuracy after one epoch:", history.history["accuracy"][0])
```

## Loss vs Epoch

This graph shows how the training and validation loss change during each epoch. As backpropagation updates the model weights, the loss should gradually decrease, indicating that the model is learning from its errors.

```python
plt.figure(figsize=(8,5))

plt.plot(history.history['loss'], label='Training Loss', marker='o')
plt.plot(history.history['val_loss'], label='Validation Loss', marker='s')

plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

## Accuracy vs Epoch

This graph shows how the training and validation accuracy improve during training. As the model learns through backpropagation, the accuracy generally increases with each epoch.

```python
plt.figure(figsize=(8,5))

plt.plot(history.history['accuracy'], label='Training Accuracy', marker='o')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy', marker='s')

plt.title("Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

## Weight Change Analysis

Backpropagation updates the weights of the neural network after each training step. The graph below compares the average value of the weights before and after training, showing how the model adjusts its parameters to improve predictions.

```python
# Calculate average weight values

initial_avg = np.mean(np.abs(initial_weights))
updated_avg = np.mean(np.abs(updated_weights))

plt.figure(figsize=(6,5))

plt.bar(
    ["Before Training", "After Training"],
    [initial_avg, updated_avg]
)

plt.title("Average Weight Values")
plt.ylabel("Average Absolute Weight")

plt.grid(axis='y')

plt.show()
```

# Epoch Analysis

An epoch represents one complete pass of the training dataset through the neural network.

In this experiment, the model is trained using different numbers of epochs to analyze their effect on model performance.

Epoch values used:
- 10
- 20
- 30
- 50

For each experiment, a new model is created and trained. The final test accuracy and test loss are recorded for comparison.

## Train Models

The model is trained using different epoch values. After each training session, the test accuracy and test loss are stored for analysis.

```python
# Create a new model
model_overfit = create_model(Adam(learning_rate=0.001))

print("\nTraining model for 50 Epochs (without Early Stopping) to observe overfitting...")

history_overfit = model_overfit.fit(
    X_train,
    y_train,
    epochs=50, # Train for a large number of epochs
    batch_size=32,
    validation_split=0.2, # Use a validation set to observe its performance
    verbose=1
)

# Evaluate the model after extended training
loss_overfit, accuracy_overfit = model_overfit.evaluate(X_test, y_test, verbose=0)

print(f"\nTest Accuracy (after 50 epochs): {accuracy_overfit:.4f}")
print(f"Test Loss (after 50 epochs)    : {loss_overfit:.4f}")
```

### Plotting Training and Validation Metrics

Now, let's visualize the training and validation loss and accuracy over these 50 epochs. Pay close attention to the validation curves.

```python
plt.figure(figsize=(12, 5))

# Plot Training & Validation Loss
plt.subplot(1, 2, 1)
plt.plot(history_overfit.history['loss'], label='Training Loss', marker='o', markersize=4, linestyle='-')
plt.plot(history_overfit.history['val_loss'], label='Validation Loss', marker='s', markersize=4, linestyle='--')
plt.title('Loss Over 50 Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Plot Training & Validation Accuracy
plt.subplot(1, 2, 2)
plt.plot(history_overfit.history['accuracy'], label='Training Accuracy', marker='o', markersize=4, linestyle='-')
plt.plot(history_overfit.history['val_accuracy'], label='Validation Accuracy', marker='s', markersize=4, linestyle='--')
plt.title('Accuracy Over 50 Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Observation:

In the plots above, you should observe the following:

*   **Training Loss** (solid blue line in the left plot) will likely continue to decrease throughout the 50 epochs, as the model gets better at fitting the training data.
*   **Validation Loss** (dashed orange line in the left plot) will initially decrease but then typically starts to **increase** after a certain number of epochs. This is the crucial sign of overfitting.
*   **Training Accuracy** (solid blue line in the right plot) will likely continue to increase or stay very high.
*   **Validation Accuracy** (dashed orange line in the right plot) will increase for some time, reach a peak, and then might **plateau or even start to decrease** as the model overfits.

This behavior visually confirms that training for an excessively high number of epochs without techniques like Early Stopping leads to a model that performs worse on unseen data, even if it seems to be doing better on the training data.

## Learning Rate Analysis

The learning rate is a crucial hyperparameter that determines the step size at each iteration while moving toward a minimum of the loss function. Choosing an appropriate learning rate is vital for efficient training.

In this section, we will train models with different learning rates to observe their impact on training and validation performance.

```python
learning_rates = [0.1, 0.01, 0.001, 0.0001]
epochs_for_lr_analysis = 20 # Using 20 epochs for consistency based on previous analysis

lr_histories = {}

for lr in learning_rates:
    print("=" * 50)
    print(f"Training Model with Learning Rate: {lr}")
    print("=" * 50)

    # Create a new model for each learning rate
    model_lr = create_model(Adam(learning_rate=lr))

    history_lr = model_lr.fit(
        X_train,
        y_train,
        epochs=epochs_for_lr_analysis,
        batch_size=32,
        validation_split=0.2,
        verbose=0 # Set to 0 to avoid verbose output during multiple trainings
    )
    lr_histories[lr] = history_lr
```

### Visualize Learning Rate Effects on Loss

These plots compare the training and validation loss for models trained with different learning rates.

```python
plt.figure(figsize=(12, 6))

for lr, history in lr_histories.items():
    plt.plot(history.history['loss'], label=f'Training Loss (LR={lr})', linestyle='-')
    plt.plot(history.history['val_loss'], label=f'Validation Loss (LR={lr})', linestyle='--')

plt.title('Loss Curves for Different Learning Rates')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

### Visualize Learning Rate Effects on Accuracy

These plots compare the training and validation accuracy for models trained with different learning rates.

```python
plt.figure(figsize=(12, 6))

for lr, history in lr_histories.items():
    plt.plot(history.history['accuracy'], label=f'Training Accuracy (LR={lr})', linestyle='-')
    plt.plot(history.history['val_accuracy'], label=f'Validation Accuracy (LR={lr})', linestyle='--')

plt.title('Accuracy Curves for Different Learning Rates')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()
```

### Observations on Learning Rate Analysis

Analyze the plots above to understand the impact of different learning rates:

*   **High Learning Rate (e.g., 0.1, 0.01):** The loss might drop very quickly, but it can also oscillate wildly or even diverge. The model might overshoot the optimal weights, leading to unstable training and poor generalization.
*   **Low Learning Rate (e.g., 0.0001):** The training might be very slow, taking many epochs to converge. While it might reach a good minimum, the process is inefficient.
*   **Optimal Learning Rate (e.g., 0.001):** This typically shows a smooth decrease in loss and a steady increase in accuracy on both training and validation sets, indicating efficient learning without much oscillation or slow convergence. This rate allows the model to find a good balance between speed and stability.

From these observations, you can select a learning rate that provides a good trade-off between training speed and model performance.
